## Data Cleaning

The following syntax is done to do an initial duplicate deletion on the entire raw dataset saved in 'df_delduplicate' (to ensure the original df is not touched).

This deletion at 23 columns, rather than doing it on the selected 8 columns, was done to avoid the removal of valid listings. This is because rows that look similar on the 8 columns may still be different listings on the 23 columns. If we had done the duplicates deletion after selecting the 8 columns into a new dataframe, these listings would be deleted and cause unreliable results during the analysis.

```python
df_delduplicate = df.drop_duplicates()
df_delduplicate.shape
```

The following syntax is to select the 8 columns that will be cleaned and used in the analysis later on, saving them into a variable named 'df_cleaned'.

As well as using .head(50) to show the first 50 rows of the df_cleaned.

```python
df_cleaned = df_delduplicate[['Price', 'District', 'Type', 'TotalArea', 'Parking', 'Elevator', 'NumberOfBathrooms', 'EnergyCertificate']].copy()

df_cleaned.head(50)
```

The following syntax is used to remove any rows that has missing values.

```python
df_cleaned = df_cleaned.dropna()
df_cleaned.shape
```

The following syntax is used to remove rows with any values that do not make sense. This is done by setting judgement cut off values or filters to only include realistic values in the dataset, and saving them into df_cleaned.

NOTE: These values were derived from personal judgements of what is sensible in a property listing. The need for this came about when previous syntaxes showed that the distribution of data of columns 'Price', 'TotalArea', and 'NumberOfBathrooms' did not make sense.

```python
df_cleaned = df_cleaned[df_cleaned['Price'] >= 1000]
df_cleaned = df_cleaned[df_cleaned['Price'] <= 60000000]

df_cleaned = df_cleaned[df_cleaned['TotalArea'] > 0]
df_cleaned = df_cleaned[df_cleaned['TotalArea'] <= 8000000]

df_cleaned = df_cleaned[df_cleaned['NumberOfBathrooms'] >= 0]
df_cleaned = df_cleaned[df_cleaned['NumberOfBathrooms'] <= 15]

df_cleaned.shape
```

The following syntax is used to remove rows with 'Z - Fora de Portugal' as a district.

```python
df_cleaned = df_cleaned[df_cleaned['District'] != 'Z - Fora de Portugal']
df_cleaned['District'].value_counts()
```

The following syntax is used to combine 'NC', 'No Certificate', and 'Not available' as 'Not Certificated'.

```python
df_cleaned['EnergyCertificate'] = df_cleaned['EnergyCertificate'].replace(['NC', 'No Certificate', 'Not available'], 'Not Certificated')
df_cleaned['EnergyCertificate'].value_counts()
```
